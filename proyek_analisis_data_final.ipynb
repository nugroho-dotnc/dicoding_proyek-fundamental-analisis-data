## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

**Contoh pertanyaan bisnis yang memenuhi seluruh elemen SMART**

***"Faktor apa saja yang memengaruhi penurunan conversion rate pada pengguna aplikasi Android di wilayah Jabodetabek sebesar 5% selama periode Flash Sale Maret 2026?"***

Keterangan:

- **Specific**: Fokus pada "penurunan conversion rate" untuk "aplikasi Android" di "Jabodetabek". Bukan sekadar penjualan turun.
- **Measurable**: Ada angka konkret yang ingin dianalisis, yaitu penurunan sebesar "5%".
- **Action-Oriented**: Dengan mengetahui faktor penyebabnya misalnya bug pada versi Android tertentu atau kendala logistik di Jabodetabek, tim bisa langsung melakukan perbaikan teknis atau operasional.
- **Relevant**: Penurunan konversi saat Flash Sale adalah masalah kritis bagi bisnis retail/e-commerce.
- **Time-bound**: Dibatasi pada periode spesifik "Flash Sale Maret 2026".

- **Pertanyaan 1:** Berapa total penjualan produk dan rata - rata tingkat kepuasan pelanggan untuk 5 kategori produk teratas di wilayah SP pada 1 juni hingga 30 agustus tahun 2018, sehingga tim marketing bisa memfokuskan anggaran promosi pada 5 kategori tersebut?
- **Pertanyaan 2:** Berapa median selisih antara tanggal estimasi dan tanggal sampai sebenarnya, berapa persen pesanan yang terlambat, dan apakah keterlambatan menurunkan rata-rata review score, untuk pesanan delivered ke wilayah SP pada 1 Januari sampai 31 Agustus 2018 (berdasarkan tanggal pesanan datang)?
- **Pertanyaan 3:** Negara bagian mana di luar SP yang memiliki rata - rata freight_value tertinggi sepanjang tahun 2017, dan bagaimana dampaknya terhadap total volume pesanan agar perusahaan dapat memutuskan untuk membuka pusat distribusi di wilayah tersebut?

cleaning data orders

```python
# terdapat 8 jenis order_status: created -> approved -> invoiced -> processing -> shipped -> delivered | 'cancelled' dan 'unavailable' bisa membuat data null kapan saja
# mengubah tipe data kolom ke dalam bentuk datetime

clean_orders_df = orders_df.copy()
order_to_datetime_columns = ['order_purchase_timestamp', 
                             'order_approved_at', 
                             'order_delivered_carrier_date', 
                             'order_delivered_customer_date', 
                             'order_estimated_delivery_date', ]

for col in order_to_datetime_columns:
    clean_orders_df[col] = pd.to_datetime(clean_orders_df[col])


# disini terlihat ada 14 empty values pada kolom order_approved_at, 2 pada order_delivered_carrier_date, dan 8 pada order_delivered_customer_date padahal order_status sudah delivered
clean_orders_df[(clean_orders_df['order_status'] == 'delivered')].isna().sum()

# cari data yang tidak sesuai alur, tambahkan flag is_timestamp_anomaly
o = clean_orders_df

# cek logika empty values per status
tanggal_cols = ['order_approved_at', 'order_delivered_carrier_date', 'order_delivered_customer_date']
display(o.groupby('order_status')[tanggal_cols].apply(lambda x: x.isna().sum()))

# status non-delivered tetapi tanggal sampai pelanggan tidak null
display(o[(o['order_status'] != 'delivered') & o['order_delivered_customer_date'].notna()].groupby('order_status').size())

# tambahkan flag anomali

# anomali yang merusak metrik pengiriman
o['is_delivery_anomaly'] = (
    (o['order_delivered_customer_date'] < o['order_delivered_carrier_date']) | (o['order_delivered_customer_date'] < o['order_purchase_timestamp'])
)

# anomali urutan proses internal
o['is_process_anomaly'] = (
    (o['order_approved_at'] < o['order_purchase_timestamp']) |
    (o['order_delivered_carrier_date'] < o['order_approved_at'])
)

# delivered tetapi tanggal sampai ke pelanggan kosong
o['is_delivered_date_missing'] = (
    (o['order_status'] == 'delivered') & (o['order_delivered_customer_date'].isna())
)

print(o[['is_delivery_anomaly', 'is_process_anomaly', 'is_delivered_date_missing']].sum())

# penambahan kolom untuk pertanyaan 2
o['total_waktu_pengiriman'] = (
    o['order_delivered_customer_date'] - o['order_purchase_timestamp']
).dt.total_seconds()/86400

o['selisih_estimasi_vs_aktual'] = (
    o['order_estimated_delivery_date'] - o['order_delivered_customer_date']
).dt.total_seconds() / 86400

# durasi hanya valid untuk delivered yang tidak anomali
valid_mask = (
    (o['order_status'] == 'delivered') & o['order_delivered_customer_date'].notna() & ~o['is_delivery_anomaly']
)

o.loc[~valid_mask, ['total_waktu_pengiriman', 'selisih_estimasi_vs_aktual']] = np.nan

# cek hasilnya
print(f"delivered valid: {valid_mask.sum()} dari {(o['order_status'] == 'delivered').sum()} order_delivered")
display(o[['total_waktu_pengiriman', 'selisih_estimasi_vs_aktual']].describe())
```

analisis  outlier pada kolom total_waktu_pengiriman dan seliih_estimasi_vs_aktual

```python
q2 = o[valid_mask].merge(
    clean_customers_df[['customer_id', 'customer_state']], on='customer_id', how='left')

q2 = q2[(q2['customer_state'] == 'SP') &
        (q2['order_purchase_timestamp'].dt.year == 2017)].copy()
print(len(q2))
```

```python
cols = ['total_waktu_pengiriman', 'selisih_estimasi_vs_aktual']
display(q2[cols].describe(percentiles=[.01, .05, .25, .5, .75, .95, .99]))

for c in cols:
    q1, q3 = q2[c].quantile([.25, .75])
    iqr = q3 - q1
    low, high = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    q2[f'outlier_{c}'] = (q2[c] < low) | (q2[c] > high)
    print(c, f'batas: {low:.1f} s/d {high:.1f}', '| outlier:', q2[f'outlier_{c}'].sum())

# visual cepat
fig, ax = plt.subplots(1, 2, figsize=(12, 4))
sns.boxplot(x=q2[cols[0]], ax=ax[0]); sns.boxplot(x=q2[cols[1]], ax=ax[1])
plt.show()
```

```python
rev = clean_order_reviews_df[['order_id', 'review_score']]
q2 = q2.merge(rev, on='order_id', how='left')

# apakah outlier terlambat punya review lebih rendah?
print(q2.groupby('outlier_selisih_estimasi_vs_aktual')['review_score'].agg(['count', 'mean']))
```

```python
# supaya  lebih aman, outlier tidak dihapus namun hanya diberi flag agar bisa di filter dengan lebih mudah
for c in ['total_waktu_pengiriman', 'selisih_estimasi_vs_aktual']:
    q1, q3 = o.loc[valid_mask, c].quantile([.25, .75])
    iqr = q3 - q1
    o[f'is_outlier_{c}'] = (o[c] < q1 - 1.5*iqr) | (o[c] > q3 + 1.5*iqr)
    print(c, o[f'is_outlier_{c}'].sum())
```
